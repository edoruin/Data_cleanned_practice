```python
import pandas as pd
import numpy as np
```

# Descriptive Statistics

```python
# Loading the datasets
test = pd.read_csv('Titanic/test.csv')
train = pd.read_csv('Titanic/train.csv')

train.head()
```

```python
# The 'Name' column is removed as it won't be used in this analysis
train.drop(['Name'], axis=1, inplace=True)
test.drop(['Name'], axis=1, inplace=True)
```

```python
# Handling missing values in categorical variables using SimpleImputer
from sklearn.impute import SimpleImputer

cat_imputer = SimpleImputer(strategy='most_frequent')

# Identifying categorical columns
cat_cols = train.select_dtypes(include='object').columns

# Applying the imputer to both training and test sets
train[cat_cols] = cat_imputer.fit_transform(train[cat_cols])
test[cat_cols] = cat_imputer.fit_transform(test[cat_cols])
```

```python
# Filling missing values in numerical columns

# Imputing 'Age' with the mean and converting it to integer
train['Age'] = train['Age'].fillna(train['Age'].mean())
train['Age'] = train['Age'].astype('int')

test['Age'] = test['Age'].fillna(test['Age'].mean())
test['Age'] = test['Age'].astype('int')
```

```python
# Applying One-Hot Encoding to categorical variables
from sklearn.preprocessing import OneHotEncoder

# Initializing the OneHotEncoder
OH_encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

# Selecting specific columns to encode: 'Sex' and 'Embarked'
cat_cols_to_encode = ['Sex', 'Embarked']

# Performing the transformation
Encoder_train = pd.DataFrame(OH_encoder.fit_transform(train[cat_cols_to_encode]))
Encoder_test = pd.DataFrame(OH_encoder.transform(test[cat_cols_to_encode]))

# Restoring indices
Encoder_train.index = train.index
Encoder_test.index = test.index

# Removing the original categorical columns
num_X_train = train.drop(cat_cols_to_encode, axis=1)
num_X_test = test.drop(cat_cols_to_encode, axis=1)

# Concatenating the encoded columns with the rest of the data
data_train = pd.concat([num_X_train, Encoder_train], axis=1)
data_test = pd.concat([num_X_test, Encoder_test], axis=1)

# Ensuring all column names are strings for consistency
data_train.columns = data_train.columns.astype(str)
data_test.columns = data_test.columns.astype(str)
```

```python
# Dropping 'Ticket' and 'Cabin' columns for now.
# These could be further processed in future iterations.
data_train = data_train.drop(['Ticket', 'Cabin'], axis=1)
data_test = data_test.drop(['Ticket', 'Cabin'], axis=1)
```

```python
# Exporting the cleaned datasets to CSV files
data_train.to_csv('V_train_cleaned.csv', index=False)
data_test.to_csv('V_test_cleaned.csv', index=False)
```
